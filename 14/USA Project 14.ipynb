```python
from ddgs import DDGS
import wikipedia

i = input("Which jet do you have in mind? ")
query = f"Technical specifications and history of {i} fighter jet"
print("Searching on Google...")
print("Trying to connect")

with DDGS() as ddgs:
    results = ddgs.text(query, max_results=5)
    
    for r in results:
        print(f"Title: {r['title']}")
        print(f"Link:{r['href']}")
        print("-" * 30)
print("\nGetting information from Wikipedia...\n")

try:
    search_results = wikipedia.search(i + " fighter jet")
    
    if search_results:
        page_title = search_results[0]
        page = wikipedia.page(page_title)

        print(f"Wikipedia Title: {page.title}")
        print(f"Wikipedia URL: {page.url}")
        print("\nSummary:")
        print(wikipedia.summary(page_title, sentences=5))
    else:
        print("No Wikipedia page found for this jet(But maybe there's something on Wikipedia itself).")

except wikipedia.exceptions.DisambiguationError as e:
    print("This name is ambiguous. Possible options:")
    for option in e.options[:5]:
        print("-", option)

except wikipedia.exceptions.PageError:
    print("No exact Wikipedia page found(But maybe there's something on Wikipedia itself.).")

except Exception as e:
    print("An error occurred:", e)
```
